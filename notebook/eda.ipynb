```python
df['TotalCharges'] = df['TotalCharges'].replace(' ', '0')
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'])
```

```python
df['ChurnFlag'] = (df['Churn'] == 'Yes').astype(int)
```

```python
df[['Churn', 'ChurnFlag']].head(10)
```

```python
def tenure_bucket(t):
    if t <= 12:
        return '0-1 yr'
    elif t <= 24:
        return '1-2 yr'
    elif t <= 48:
        return '2-4 yr'
    else:
        return '4+ yr'

df['TenureBucket'] = df['tenure'].apply(tenure_bucket)
```

```python
df[['tenure', 'TenureBucket']].sample(10)
```

```python
service_cols = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

df['NumServices'] = df[service_cols].apply(lambda row: sum(v == 'Yes' for v in row), axis=1)
```

```python
df[service_cols + ['NumServices']].sample(10)
```
